```python
#palindrome
def palindrome(n):
  original = n
  rev = 0
  while(n!=0):
    rem=n%10
    rev=rev*10+rem
    n=n//10
  if(rev==original):
    return True
  else:
    return False

print(palindrome(252))
print(palindrome(645))

```

```python
#reverse the string
def reverse_str(a):
  rev = ""
  for i in range(len(a)-1,-1,-1):
    rev = rev+a[i]
  return rev
print(reverse_str("afrin"))

```

```python
#example on default argument
def default_args(name,branch="AI&ML"):
  print("my name",name,"my branch",branch)
print (default_args("afrin"))
```

```python
# example on keyword
def keyword_args(name,branch):
  print("my name",name,"my branch",branch)

print(keyword_args(name="afrin",branch = "AI&DS"))
```

```python
#variable length argument
def mult(*num):
  print(num)
mult(1,2,4,6.7)
```

```python
#multiple v alues in function
def mult_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=mult_values(1,2,3,4,5)
print(f"sum={res_sum},count={res_count},the average is {res_avg}")
```
