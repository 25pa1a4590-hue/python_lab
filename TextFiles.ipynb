```python
#creting and reading  a text data
f=open("data.txt","w")
content="My name is afrin"
f.write(content)
print("file created")
f.close()
```

```python
#to check wheather file is closed
f.closed
```

```python
#to check the name of the file
f.name
```

```python
#to check the mode
f.mode
```

```python
#to append data
f=open("data.txt","a")
#by using writelines we can give data in list format
f.writelines(["\ncurrently pursuing b.tech","\nIn VIT"])
f.close()
```

```python
#to seek and tell
f=open("data.txt","r")
content=f.read()
print(content)
print(f.tell())
f.seek(2)
print(f.read())
```

```python
#to print data in list format
f=open("data.txt","r")
content=f.readlines()
print(content)
```
