```python
#operations on tuples
t1=(7,8,6)
t2=(3,4,5)
print(t1[2]) #indexing
print(t1+t2) #concatenation
print(t1[0:1]) #slicing
print(t2*3) #repetition
print(7 in t1) #membership
for i in t1: #iteration
  print(i)
```

```python
#code on functions in tuple
t3=(9,10,11)
print(sum(t3))
print(max(t3))
print(min(t3))
print(len(t3))
```

```python
#code on methods in tuple
t4=(11,22,22,33,'python')
print(t4.count(22))
print(t4.index('python'))
```

```python
#code on zip in tuple
t5=(22,55,66)
t6=(33,44,77)
print(tuple(zip(t5,t6)))
print(list(zip(t5,t6)))
```

```python
#code on to count no of vowels
t3=str(input("enter a string:"))
t3=t3.lower()
vowels_count=t3.count('a')+t3.count('e')+t3.count('i')+t3.count('o')+t3.count('u')
print("the no of vowels are:",vowels_count)
```
